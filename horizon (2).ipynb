### Critical Feature Leakage Advisory
A vital distinction in professional data science is separating **pre-churn operational indicators** from **post-event diagnostic fields**:
- `Satisfaction Score` (1–5) and `Churn Score` (0–100) are recorded upon customer exit or derived directly from customer defection indicators ($r = -0.752$ and $r = +0.660$).
- `Churn Category` and `Churn Reason` exist *exclusively* for departed accounts.
- `Customer Status` ('Stayed', 'Churned', 'Joined') directly duplicates the target variable.
*Methodological Note: While valuable for post-mortem diagnostics, these fields represent post-event leakage and must be strictly excluded from independent operational driver identification in PS7.*

## PS1: Data Audit & Quality Cleaning Pipeline

### Objective and Purpose of Data Audit
Before undertaking analytical exploration, raw transactional data must undergo a rigorous, documented data quality audit. Operational datasets frequently suffer from datatype mismatches (such as currency strings stored as text), hidden duplicate entries, and ambiguous missing-value structures. 

The audit pipeline evaluates:
1. **Dataset Structure and Schema Integrity**: Verifying that every column conforms to its expected numerical or categorical representation.
2. **Duplicate Records & Primary Key Uniqueness**: Validating that `Customer ID` uniquely indexes each record with zero duplicated customer sessions.
3. **Numeric String Sanitization (`pd.to_numeric`)**: Converting financial charges, usage metrics, and tenure fields into high-precision floating-point or integer types.
4. **Missing-Value Mechanism Audit**: Distinguishing between random missingness (data collection failure) and structural missingness (valid business skips).
5. **Domain-Grounded Imputation Decisions**: Replacing structural missingness with exact domain labels without distorting underlying distributions.
6. **Cleaning Log & Verification**: Recording all transformations and validating zero data leakage or unresolved NaNs.
7. **Cleaned Dataset Export**: Storing the validated dataframe for seamless reproducibility across subsequent problem statements.

<h2 style="background-color:#154360; color:#FFFFFF; padding:10px;">
PS1: Data Audit & Quality Cleaning Pipeline
</h2>

````python

cleaning_log = []
cleaning_log.append(f"Ingested raw dataset shape: {df_raw.shape[0]} rows, {df_raw.shape[1]} columns.")

# Step 1: Duplicate Detection & Identifier Integrity
full_row_dups = df_raw.duplicated().sum()
id_dups = df_raw['Customer ID'].duplicated().sum() if 'Customer ID' in df_raw.columns else 0
cleaning_log.append(f"Duplicate Check: Found {full_row_dups} full row duplicates and {id_dups} Customer ID duplicates.")

df_clean = df_raw.drop_duplicates().copy()
if 'Customer ID' in df_clean.columns:
    df_clean = df_clean.drop_duplicates(subset=['Customer ID'], keep='first')

# Step 2: Numeric Type Conversion (pd.to_numeric)
numeric_target_cols = [
    'Age', 'Avg Monthly GB Download', 'Avg Monthly Long Distance Charges',
    'Churn Score', 'CLTV', 'Dependents', 'Device Protection Plan',
    'Internet Service', 'Latitude', 'Longitude', 'Married', 'Monthly Charge',
    'Multiple Lines', 'Number of Dependents', 'Number of Referrals',
    'Online Backup', 'Online Security', 'Paperless Billing', 'Partner',
    'Phone Service', 'Population', 'Premium Tech Support', 'Referred a Friend',
    'Satisfaction Score', 'Senior Citizen', 'Streaming Movies', 'Streaming Music',
    'Streaming TV', 'Tenure in Months', 'Total Charges', 'Total Extra Data Charges',
    'Total Long Distance Charges', 'Total Refunds', 'Total Revenue', 'Under 30',
    'Unlimited Data', 'Zip Code', 'Churn'
]

converted_count = 0
for col in numeric_target_cols:
    if col in df_clean.columns:
        if df_clean[col].dtype == object or str(df_clean[col].dtype).startswith('str'):
            df_clean[col] = df_clean[col].astype(str).str.replace('$', '', regex=False).str.replace(',', '', regex=False).str.strip()
        df_clean[col] = pd.to_numeric(df_clean[col], errors='coerce')
        converted_count += 1

cleaning_log.append(f"Numeric Conversion: Converted {converted_count} columns using pd.to_numeric.")

# Step 3: Missing-Value Structural Analysis & Domain Imputation
missing_before = df_clean.isna().sum()[df_clean.isna().sum() > 0]
cleaning_log.append(f"Missing Values Identified: {missing_before.to_dict()}")

# Impute domain-appropriate structural defaults:
# - Internet Type: NaN occurs when Internet Service == 0 -> 'No Internet'
# - Offer: NaN occurs when no special promo offer was chosen -> 'No Offer'
# - Churn Category: NaN occurs when customer stayed (Churn == 0) -> 'Not Churned'
# - Churn Reason: NaN occurs when customer stayed (Churn == 0) -> 'Not Churned'
df_clean['Internet Type'] = df_clean['Internet Type'].fillna('No Internet')
df_clean['Offer'] = df_clean['Offer'].fillna('No Offer')
df_clean['Churn Category'] = df_clean['Churn Category'].fillna('Not Churned')
df_clean['Churn Reason'] = df_clean['Churn Reason'].fillna('Not Churned')

remaining_nans = int(df_clean.isna().sum().sum())
cleaning_log.append(f"Domain Imputation Completed: Total unresolved NaNs remaining = {remaining_nans}.")

# Step 4: Export Cleaned Dataset for Downstream Reproducibility
cleaned_export_path = 'cleaned_customer_churn.csv'
df_clean.to_csv(cleaned_export_path, index=False)
cleaning_log.append(f"Exported validated dataset to '{cleaned_export_path}' with shape {df_clean.shape}.")

# Display Execution Summary Log
print("=" * 76)
print("PS1: COMPREHENSIVE DATA QUALITY & AUDIT VERIFICATION LOG")
print("=" * 76)
for entry in cleaning_log:
    print(f"  * {entry}")
print("=" * 76)
print(f"Final Data Integrity Check: {df_clean.shape[0]:,} rows, {df_clean.shape[1]} columns, {remaining_nans} NaNs.")
print("=" * 76)
````

PS2: Baseline Metric Layer & Class Balance Analysis
</h2>



### Business Question
What is the macro baseline attrition rate across the customer portfolio, and what is the exact structural class balance between retained and churned customer accounts?

### Why This Matters
Establishing an unassailable baseline churn metric is essential for evaluating retention health. In recurring revenue businesses, the overall churn rate governs customer replacement volume and Customer Lifetime Value (CLTV). Furthermore, measuring class balance determines the appropriate analytical perspective: evaluating performance requires balancing percentage probabilities against absolute customer counts to correctly size business retention operations.

### Analytical Method
1. Map customer `Churn` status into standard binary numerical indicators ($Y \in \{0, 1\}$).
2. Calculate core portfolio parameters:
   - Total Portfolio Volume ($N$)
   - Absolute Retained Customer Count ($N_{retained}$)
   - Absolute Churned Customer Count ($N_{churn}$)
   - Baseline Churn Rate: $\text{Churn Rate} = \frac{N_{churn}}{N} \times 100$
   - Baseline Retention Rate: $\text{Retention Rate} = \frac{N_{retained}}{N} \times 100$
   - Class Imbalance Ratio: $\text{Ratio} = 1 : \frac{N_{retained}}{N_{churn}}$
3. Construct a high-resolution, two-figure visual architecture:
   - **Plot 1**: Annotated Bar Chart displaying exact customer counts with percentage overlays.
   - **Plot 2**: Custom Donut Chart illustrating proportional share with an exploded slice and central KPI callout.

````python
# PS2: Baseline Metric Layer & Class Balance Analysis
# Ensure Churn target is mapped to standard binary integer format (1 = Churned, 0 = Retained)
if df_clean['Churn'].dtype == object or str(df_clean['Churn'].dtype).startswith('str'):
    churn_mapping = {'Yes': 1, 'No': 0, 'Churned': 1, 'Stayed': 0, 'Joined': 0, '1': 1, '0': 0}
    df_clean['Churn'] = df_clean['Churn'].map(churn_mapping)
df_clean['Churn'] = df_clean['Churn'].astype(int)

# Core Metric Formulations
total_customers = len(df_clean)
churned_customers = int(df_clean['Churn'].sum())
retained_customers = total_customers - churned_customers
churn_rate_pct = (churned_customers / total_customers) * 100
retention_rate_pct = (retained_customers / total_customers) * 100
imbalance_ratio = retained_customers / churned_customers

# Formatted Baseline Metric Scorecard
print("=" * 70)
print("PS2: BASELINE METRIC LAYER & CLASS BALANCE SUMMARY")
print("=" * 70)
print(f"Total Customer Portfolio Count (N) : {total_customers:,}")
print(f"Total Retained Customers (0)       : {retained_customers:,} ({retention_rate_pct:.2f}%)")
print(f"Total Churned Customers (1)        : {churned_customers:,} ({churn_rate_pct:.2f}%)")
print(f"Class Imbalance Ratio              : 1 : {imbalance_ratio:.2f} (1 Churned per {imbalance_ratio:.2f} Retained)")
print("=" * 70)

categories = ['Retained Customers (0)', 'Churned Customers (1)']
counts = [retained_customers, churned_customers]
palette_colors = [BI_PALETTE['retained'], BI_PALETTE['churn']]

# Figure 1: Plot 1 — Annotated Bar Chart
fig1, ax1 = plt.subplots(figsize=(8, 5.2))
bars = ax1.bar(categories, counts, color=palette_colors, width=0.48, edgecolor='black', linewidth=1.0)
for bar in bars:
    height = bar.get_height()
    pct = (height / total_customers) * 100
    ax1.annotate(f'{height:,}\n({pct:.2f}%)',
                 xy=(bar.get_x() + bar.get_width() / 2, height),
                 xytext=(0, 6), textcoords='offset points',
                 ha='center', va='bottom', fontsize=11, fontweight='bold', color='#1a252f')

ax1.set_title('Plot 1: Absolute Customer Volume by Churn Status', fontsize=13, fontweight='bold', pad=14, color=BI_PALETTE['primary'])
ax1.set_ylabel('Customer Count', fontsize=11, fontweight='bold')
ax1.set_ylim(0, max(counts) * 1.18)
ax1.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

# Figure 2: Plot 2 — Custom Donut Chart
fig2, ax2 = plt.subplots(figsize=(7, 5.2))
wedges, texts, autotexts = ax2.pie(
    counts,
    labels=categories,
    autopct='%1.2f%%',
    startangle=140,
    colors=palette_colors,
    explode=(0, 0.08),
    wedgeprops=dict(width=0.40, edgecolor='white', linewidth=2.5),
    pctdistance=0.76
)
plt.setp(autotexts, size=11, weight='bold', color='white')
plt.setp(texts, size=11, weight='bold', color='#2c3e50')

# Central Metric Callout in Donut Hole
ax2.text(0, 0, f'Baseline Churn\n{churn_rate_pct:.2f}%', ha='center', va='center',
         fontsize=13, fontweight='bold', color=BI_PALETTE['primary'])
ax2.set_title('Plot 2: Proportional Share of Retained vs. Churned', fontsize=13, fontweight='bold', pad=14, color=BI_PALETTE['primary'])
plt.tight_layout()
plt.show()
````

## PS3: Customer Lifecycle & Tenure Cohort Risk

### Business Question
At what stage of the customer lifecycle does attrition peak, and does customer retention exhibit stabilization as customer tenure deepens?

### Why This Matters
Customer acquisition costs (CAC) in telecommunications often require 12 to 18 months of recurring service payments to break even. If customer departures are heavily concentrated in the early lifecycle months ("infant mortality"), marketing and sales capital is effectively destroyed before generating positive return on investment. Locating the exact timing of churn allows customer success teams to front-load retention interventions.

### Analytical Method
1. Discretize `Tenure in Months` into six contiguous 12-month annual cohorts:
   - Cohort 1: `0-12 months` (Early Onboarding)
   - Cohort 2: `13-24 months` (Post-Onboarding)
   - Cohort 3: `25-36 months` (Mid-Tenure)
   - Cohort 4: `37-48 months` (Mature Accounts)
   - Cohort 5: `49-60 months` (Established Accounts)
   - Cohort 6: `61-72 months` (Veteran Loyalists)
2. Formulate metrics:
   - Cohort Volume ($N_c$)
   - Cohort Churned Count ($N_{c, churn}$)
   - Normalized Cohort Churn Rate: $\text{Churn Rate}_c = \frac{N_{c, churn}}{N_c} \times 100$
   - Cohort Share of Total Company Churn: $\text{Share} = \frac{N_{c, churn}}{N_{total, churn}} \times 100$
3. Render dual visual perspectives:
   - **Figure 1 (Plots 3 & 4)**: Dual-Axis Combo Plot displaying customer volume bars overlaid with a normalized churn rate curve and a shaded onboarding risk zone.
   - **Figure 2 (Plot 5)**: Horizontal Bar Chart showing each cohort's contribution to total company churn.

````python
# PS3: Customer Lifecycle & Tenure Cohort Risk Analysis
bins = [0, 12, 24, 36, 48, 60, 72]
labels = ['0-12 months', '13-24 months', '25-36 months', '37-48 months', '49-60 months', '61-72 months']
df_clean['Tenure_Cohort'] = pd.cut(df_clean['Tenure in Months'], bins=bins, labels=labels, include_lowest=True)

tenure_cohorts = df_clean.groupby('Tenure_Cohort', observed=False).agg(
    Total_Customers=('Churn', 'count'),
    Retained_Customers=('Churn', lambda x: (x == 0).sum()),
    Churned_Customers=('Churn', 'sum')
)
tenure_cohorts['Churn_Rate_Pct'] = (tenure_cohorts['Churned_Customers'] / tenure_cohorts['Total_Customers']) * 100
tenure_cohorts['Share_of_Total_Churn_Pct'] = (tenure_cohorts['Churned_Customers'] / churned_customers) * 100

print("=" * 88)
print("PS3: TENURE COHORT RISK & CUSTOMER LIFECYCLE BREAKDOWN")
print("=" * 88)
print(tenure_cohorts.round(2).to_string())
print("=" * 88)

# Figure 1: Dual-Axis Combo Plot (Plots 3 & 4)
fig1, ax1 = plt.subplots(figsize=(13, 6))
x = np.arange(len(labels))
width = 0.52

# Plot 3: Stacked Bar Chart for Customer Volume
ax1.bar(x, tenure_cohorts['Retained_Customers'], width, label='Retained Customers',
        color=BI_PALETTE['retained'], alpha=0.88, edgecolor='black', linewidth=0.8)
ax1.bar(x, tenure_cohorts['Churned_Customers'], width, bottom=tenure_cohorts['Retained_Customers'],
        label='Churned Customers', color=BI_PALETTE['churn'], alpha=0.92, edgecolor='black', linewidth=0.8)

for i, total in enumerate(tenure_cohorts['Total_Customers']):
    ax1.annotate(f'N={total:,}', xy=(i, total), xytext=(0, 6), textcoords='offset points',
                 ha='center', va='bottom', fontsize=10, fontweight='bold', color='#2c3e50')

ax1.set_xlabel('Customer Tenure Cohort (Lifecycle Stage)', fontsize=11, fontweight='bold', labelpad=10)
ax1.set_ylabel('Cohort Customer Volume', fontsize=11, fontweight='bold', color=BI_PALETTE['retained'])
ax1.set_xticks(x)
ax1.set_xticklabels(labels, fontsize=10, fontweight='bold')
ax1.set_ylim(0, max(tenure_cohorts['Total_Customers']) * 1.20)
ax1.grid(axis='y', alpha=0.3)

# Plot 4: Normalized Churn Rate Line Overlay
ax2 = ax1.twinx()
line = ax2.plot(x, tenure_cohorts['Churn_Rate_Pct'], color='#e74c3c', marker='o',
                linewidth=3.2, markersize=9, label='Cohort Churn Rate (%)')
ax2.set_ylabel('Normalized Churn Rate (%)', fontsize=11, fontweight='bold', color='#e74c3c')
ax2.set_ylim(0, 60)
ax2.grid(False)

for i, val in enumerate(tenure_cohorts['Churn_Rate_Pct']):
    ax2.annotate(f'{val:.1f}%', xy=(i, val), xytext=(0, 10), textcoords='offset points',
                 ha='center', va='bottom', fontsize=10, fontweight='bold', color='#c0392b',
                 bbox=dict(boxstyle='round,pad=0.25', facecolor='white', edgecolor='#e74c3c', alpha=0.95))

# Shaded background highlight over the Critical Onboarding Window
ax1.axvspan(-0.35, 0.35, color='#ffeaa7', alpha=0.45, linestyle=':', label='Critical Onboarding Zone (0-12m)')

# Unified legend
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc='upper right', frameon=True, framealpha=0.92)
plt.title('Plots 3 & 4: Customer Lifecycle Dynamics — Volume vs. Normalized Churn Rate (%)', fontsize=13, fontweight='bold', pad=15, color=BI_PALETTE['primary'])
plt.tight_layout()
plt.show()

# Figure 2: Plot 5 — Cohort Share of Total Company Churn
fig2, ax_churn = plt.subplots(figsize=(11, 4.5))
bars_share = ax_churn.barh(labels[::-1], tenure_cohorts['Share_of_Total_Churn_Pct'].values[::-1],
                           color=[BI_PALETTE['safe'] if s < 10 else (BI_PALETTE['warning'] if s < 20 else BI_PALETTE['churn']) for s in tenure_cohorts['Share_of_Total_Churn_Pct'].values[::-1]],
                           edgecolor='black', linewidth=0.8, height=0.55)

for bar in bars_share:
    w = bar.get_width()
    ax_churn.annotate(f'{w:.1f}% ({int(w * churned_customers / 100):,} accounts)',
                      xy=(w + 0.8, bar.get_y() + bar.get_height() / 2),
                      va='center', fontsize=10, fontweight='bold', color='#2c3e50')

ax_churn.set_title('Plot 5: Tenure Cohort Contribution to Total Churned Customers (%)', fontsize=12, fontweight='bold', pad=12, color=BI_PALETTE['primary'])
ax_churn.set_xlabel('Share of Total Churned Population (%)', fontsize=10, fontweight='bold')
ax_churn.set_xlim(0, 65)
ax_churn.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.show()
````

### Business Question
How do contractual commitment lengths, payment processing mechanisms, and paperless billing modes interact to expose the business to elevated customer attrition?

### Why This Matters
Contract duration defines legal commitment and switching barriers, whereas payment and billing channels dictate customer friction. Payment methods requiring active customer intervention (such as checks or manual bank deductions) frequently fail, triggering involuntary churn. Identifying high-risk commercial setups provides operations teams with clear targets for contract migration and billing optimization.

### Analytical Method
1. Segment `Churn` across three operational dimensions:
   - **Contract Type**: `Month-to-Month`, `One Year`, `Two Year`
   - **Payment Method**: `Bank Withdrawal`, `Credit Card`, `Mailed Check`
   - **Paperless Billing**: `Paper Billing (0)`, `Paperless Billing (1)`
2. Cross-tabulate bivariate and trivariate combinations to isolate compound risk nexuses.
3. Compute exact segment sample sizes ($N$) alongside normalized churn percentages to maintain statistical rigor.
4. Render a unified 4-panel visual architecture:
   - **Plot 6**: Stacked 100% normalized bar chart of Contract Type.
   - **Plot 7**: Stacked 100% normalized bar chart of Payment Method.
   - **Plot 8**: Heatmap of Contract Type × Payment Method (annotated with churn rates and cell sample counts).
   - **Plot 9**: Heatmap of Contract Type × Paperless Billing (annotated with churn rates and cell sample counts).

````python
# PS4: Contract & Billing Operational Vulnerability Analysis
contract_ct = df_clean.groupby('Contract').agg(
    Total=('Churn', 'count'),
    Churned=('Churn', 'sum'),
    Churn_Rate_Pct=('Churn', lambda x: x.mean() * 100)
).sort_values('Churn_Rate_Pct', ascending=False)

payment_ct = df_clean.groupby('Payment Method').agg(
    Total=('Churn', 'count'),
    Churned=('Churn', 'sum'),
    Churn_Rate_Pct=('Churn', lambda x: x.mean() * 100)
).sort_values('Churn_Rate_Pct', ascending=False)

paperless_ct = df_clean.groupby('Paperless Billing').agg(
    Total=('Churn', 'count'),
    Churned=('Churn', 'sum'),
    Churn_Rate_Pct=('Churn', lambda x: x.mean() * 100)
)
paperless_ct.index = ['Paper Billing (0)', 'Paperless Billing (1)']

heatmap_matrix1 = pd.crosstab(df_clean['Contract'], df_clean['Payment Method'], values=df_clean['Churn'], aggfunc='mean') * 100
counts_matrix1 = pd.crosstab(df_clean['Contract'], df_clean['Payment Method'])

heatmap_matrix2 = pd.crosstab(df_clean['Contract'], df_clean['Paperless Billing'].map({0: 'Paper Billing', 1: 'Paperless Billing'}), values=df_clean['Churn'], aggfunc='mean') * 100
counts_matrix2 = pd.crosstab(df_clean['Contract'], df_clean['Paperless Billing'].map({0: 'Paper Billing', 1: 'Paperless Billing'}))

# Annotated matrix strings combining Churn Rate and Sample Size
annot_matrix1 = heatmap_matrix1.round(1).astype(str) + "%\n(N=" + counts_matrix1.astype(str) + ")"
annot_matrix2 = heatmap_matrix2.round(1).astype(str) + "%\n(N=" + counts_matrix2.astype(str) + ")"

print("=" * 80)
print("PS4: CONTRACT & BILLING OPERATIONAL VULNERABILITY SUMMARY")
print("=" * 80)
print("Contract Term Churn Breakdown:\n", contract_ct.round(2).to_string())
print("\nPayment Method Churn Breakdown:\n", payment_ct.round(2).to_string())
print("\nPaperless Billing Churn Breakdown:\n", paperless_ct.round(2).to_string())
print("=" * 80)

# Multi-panel visualization (4 panels: Plots 6, 7, 8, 9)
fig, axes = plt.subplots(2, 2, figsize=(16, 11))
plt.subplots_adjust(top=0.92, bottom=0.08, left=0.08, right=0.95, hspace=0.35, wspace=0.28)

# Plot 6: Stacked Bar Chart for Contract
ax1 = axes[0, 0]
contract_prop = pd.crosstab(df_clean['Contract'], df_clean['Churn'], normalize='index') * 100
ax1.bar(contract_prop.index, contract_prop[0], label='Retained', color=BI_PALETTE['retained'], alpha=0.88, width=0.48)
ax1.bar(contract_prop.index, contract_prop[1], bottom=contract_prop[0], label='Churned', color=BI_PALETTE['churn'], alpha=0.92, width=0.48)
for i, contract in enumerate(contract_prop.index):
    ret_pct = contract_prop.loc[contract, 0]
    churn_pct = contract_prop.loc[contract, 1]
    n_tot = len(df_clean[df_clean['Contract'] == contract])
    ax1.text(i, ret_pct / 2, f'{ret_pct:.1f}%', ha='center', va='center', color='white', fontweight='bold', fontsize=10)
    if churn_pct > 3:
        ax1.text(i, ret_pct + churn_pct / 2, f'{churn_pct:.1f}%', ha='center', va='center', color='white', fontweight='bold', fontsize=10)
    ax1.annotate(f'N={n_tot:,}', xy=(i, 102), ha='center', fontsize=10, fontweight='bold', color='#333333')
ax1.set_title('Plot 6: Churn Proportion by Contract Type', fontsize=12, fontweight='bold', pad=10, color=BI_PALETTE['primary'])
ax1.set_ylabel('Proportion of Customers (%)', fontsize=10, fontweight='bold')
ax1.set_ylim(0, 114)
ax1.legend(loc='lower left')
ax1.grid(axis='y', alpha=0.3)

# Plot 7: Stacked Bar Chart for Payment Method
ax2 = axes[0, 1]
pm_prop = pd.crosstab(df_clean['Payment Method'], df_clean['Churn'], normalize='index') * 100
ax2.bar(pm_prop.index, pm_prop[0], label='Retained', color=BI_PALETTE['retained'], alpha=0.88, width=0.48)
ax2.bar(pm_prop.index, pm_prop[1], bottom=pm_prop[0], label='Churned', color=BI_PALETTE['churn'], alpha=0.92, width=0.48)
for i, pm in enumerate(pm_prop.index):
    ret_pct = pm_prop.loc[pm, 0]
    churn_pct = pm_prop.loc[pm, 1]
    n_tot = len(df_clean[df_clean['Payment Method'] == pm])
    ax2.text(i, ret_pct / 2, f'{ret_pct:.1f}%', ha='center', va='center', color='white', fontweight='bold', fontsize=10)
    if churn_pct > 3:
        ax2.text(i, ret_pct + churn_pct / 2, f'{churn_pct:.1f}%', ha='center', va='center', color='white', fontweight='bold', fontsize=10)
    ax2.annotate(f'N={n_tot:,}', xy=(i, 102), ha='center', fontsize=10, fontweight='bold', color='#333333')
ax2.set_title('Plot 7: Churn Proportion by Payment Method', fontsize=12, fontweight='bold', pad=10, color=BI_PALETTE['primary'])
ax2.set_ylabel('Proportion of Customers (%)', fontsize=10, fontweight='bold')
ax2.set_ylim(0, 114)
ax2.legend(loc='lower left')
ax2.grid(axis='y', alpha=0.3)

# Plot 8: Heatmap Contract x Payment Method
ax3 = axes[1, 0]
sns.heatmap(heatmap_matrix1, annot=annot_matrix1, fmt='', cmap='YlOrRd', cbar_kws={'label': 'Churn Rate (%)'},
            ax=ax3, vmin=0, vmax=60, linewidths=0.6, annot_kws={'size': 10, 'weight': 'bold'})
ax3.set_title('Plot 8: Churn Rate (%): Contract × Payment Method', fontsize=12, fontweight='bold', pad=10, color=BI_PALETTE['primary'])
ax3.set_xlabel('Payment Method', fontsize=10, fontweight='bold')
ax3.set_ylabel('Contract Type', fontsize=10, fontweight='bold')

# Plot 9: Heatmap Contract x Paperless Billing
ax4 = axes[1, 1]
sns.heatmap(heatmap_matrix2, annot=annot_matrix2, fmt='', cmap='YlOrRd', cbar_kws={'label': 'Churn Rate (%)'},
            ax=ax4, vmin=0, vmax=60, linewidths=0.6, annot_kws={'size': 10, 'weight': 'bold'})
ax4.set_title('Plot 9: Churn Rate (%): Contract × Paperless Billing', fontsize=12, fontweight='bold', pad=10, color=BI_PALETTE['primary'])
ax4.set_xlabel('Billing Mode', fontsize=10, fontweight='bold')
ax4.set_ylabel('Contract Type', fontsize=10, fontweight='bold')

plt.suptitle('Plots 6-9: Contract & Billing Operational Vulnerability Architecture', fontsize=14, fontweight='bold', color=BI_PALETTE['primary'])
plt.show()
````

PS5: Product Portfolio & Service Add-on Impact Analysis
</h2>



### Business Question
Which core internet connectivity types and value-added service add-ons function as defensive retention anchors versus drivers of customer attrition?

### Why This Matters
Broadband service is often commoditized; product stickiness is created through value-added add-ons that embed into a household's daily digital life. By identifying which add-on services exhibit the largest empirical risk reduction, product managers can optimize bundling strategies to protect vulnerable accounts.

### Analytical Method
1. Segment `Churn` across core **Internet Service Types**: `Fiber Optic`, `Cable`, `DSL`, `No Internet`.
2. Evaluate 6 major **Service Add-ons**:
   - `Online Security`
   - `Premium Tech Support`
   - `Online Backup`
   - `Device Protection Plan`
   - `Streaming Movies`
   - `Streaming TV`
3. For each add-on, calculate:
   - Churn Rate for Customers WITHOUT Service: $\text{Churn}_{without}$
   - Churn Rate for Customers WITH Service: $\text{Churn}_{with}$
   - **Risk Reduction Delta (% points)**: $\Delta = \text{Churn}_{without} - \text{Churn}_{with}$
   - **Relative Risk Reduction (%)**: $\text{RRR} = \frac{\Delta}{\text{Churn}_{without}} \times 100$
4. Render visual comparisons:
   - **Plot 10**: Horizontal Comparative Bar Chart ranking add-ons by Risk Reduction Delta with sample sizes and positive/negative color coding.
   - **Plot 11**: Bar Chart of Churn Rate by Internet Service Infrastructure Type.

````python
# PS5: Product Portfolio & Service Add-on Impact Analysis
addons_meta = [
    ('Online Security', 'Online Security'),
    ('Premium Tech Support', 'Premium Tech Support'),
    ('Online Backup', 'Online Backup'),
    ('Device Protection Plan', 'Device Protection'),
    ('Streaming Movies', 'Streaming Movies'),
    ('Streaming TV', 'Streaming TV')
]

addon_results = []
for col_name, display_name in addons_meta:
    n_with = int((df_clean[col_name] == 1).sum())
    n_without = int((df_clean[col_name] == 0).sum())
    with_churn = df_clean[df_clean[col_name] == 1]['Churn'].mean() * 100
    without_churn = df_clean[df_clean[col_name] == 0]['Churn'].mean() * 100
    delta = without_churn - with_churn
    rel_red = (delta / without_churn) * 100
    addon_results.append({
        'Service Add-on': display_name,
        'Subscribers (With)': n_with,
        'Non-Subscribers (Without)': n_without,
        'Churn Rate With (%)': with_churn,
        'Churn Rate Without (%)': without_churn,
        'Risk Reduction Delta (% pts)': delta,
        'Relative Risk Reduction (%)': rel_red
    })

addon_df = pd.DataFrame(addon_results).sort_values('Risk Reduction Delta (% pts)', ascending=True)

# Internet Service Type Breakdown
internet_df = df_clean.groupby('Internet Type').agg(
    Total=('Churn', 'count'),
    Churned=('Churn', 'sum'),
    Churn_Rate_Pct=('Churn', lambda x: x.mean() * 100)
).sort_values('Churn_Rate_Pct', ascending=False)

print("=" * 96)
print("PS5: PRODUCT PORTFOLIO & SERVICE ADD-ON IMPACT ANALYSIS")
print("=" * 96)
print("Service Add-on Protective Value Ranking (Sorted by Risk Reduction Delta):")
print(addon_df.sort_values('Risk Reduction Delta (% pts)', ascending=False).round(2).to_string(index=False))
print("\nInternet Connection Infrastructure Risk Profile:")
print(internet_df.round(2).to_string())
print("=" * 96)

# Figure 1: Plot 10 — Horizontal Bar Chart of Add-on Protection
fig_addon, ax1 = plt.subplots(figsize=(10, 5.2))
bars_colors = [BI_PALETTE['safe'] if d > 0 else BI_PALETTE['churn'] for d in addon_df['Risk Reduction Delta (% pts)']]
bars = ax1.barh(addon_df['Service Add-on'], addon_df['Risk Reduction Delta (% pts)'],
                color=bars_colors, edgecolor='black', linewidth=0.8, height=0.52)
ax1.axvline(0, color='black', linewidth=1.2, linestyle='--')

for bar in bars:
    w = bar.get_width()
    xpos = w + (0.5 if w >= 0 else -0.5)
    ha = 'left' if w >= 0 else 'right'
    sign = '+' if w > 0 else ''
    ax1.annotate(f'{sign}{w:.2f}% pts', xy=(xpos, bar.get_y() + bar.get_height() / 2),
                 va='center', ha=ha, fontsize=10, fontweight='bold', color='#2c3e50')

ax1.set_title('Plot 10: Protective Value of Service Add-ons\n(Risk Reduction Delta = Without% - With%)', fontsize=12, fontweight='bold', pad=12, color=BI_PALETTE['primary'])
ax1.set_xlabel('Risk Reduction Delta (Percentage Points)', fontsize=10, fontweight='bold')
ax1.set_xlim(-9, 23)
ax1.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.show()

# Figure 2: Plot 11 — Churn by Internet Service Type
fig_int, ax2 = plt.subplots(figsize=(8, 5.2))
int_colors = [BI_PALETTE['churn'], '#7570b3', BI_PALETTE['retained'], BI_PALETTE['safe']]
bars2 = ax2.bar(internet_df.index, internet_df['Churn_Rate_Pct'], color=int_colors, width=0.48, edgecolor='black', linewidth=0.8)
for i, bar in enumerate(bars2):
    h = bar.get_height()
    tot_cnt = internet_df.iloc[i]['Total']
    churn_cnt = internet_df.iloc[i]['Churned']
    ax2.annotate(f'{h:.2f}%\n({churn_cnt:,}/{tot_cnt:,})',
                 xy=(bar.get_x() + bar.get_width() / 2, h),
                 xytext=(0, 5), textcoords='offset points',
                 ha='center', va='bottom', fontsize=10, fontweight='bold')

ax2.set_title('Plot 11: Churn Rate by Internet Service Infrastructure', fontsize=12, fontweight='bold', pad=12, color=BI_PALETTE['primary'])
ax2.set_ylabel('Cohort Churn Rate (%)', fontsize=10, fontweight='bold')
ax2.set_ylim(0, 50)
ax2.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
````

#### Figure Interpretation

- **Plot 11 (Internet Connection Infrastructure)** reveals that **Fiber Optic** accounts experience an acute **40.08% churn rate**, compared to Cable (**24.70%**), DSL (**19.84%**), and No Internet (**7.22%**).

#### Evidence and Findings
1. **Utility & Security as Retention Anchors**:
   - **Online Security**: Churn rate is **14.16% with** vs. **31.65% without** -> **+17.49 percentage points Risk Reduction Delta** (**55.26% relative risk reduction**).
   - **Premium Tech Support**: Churn rate is **15.00% with** vs. **31.39% without** -> **+16.39 percentage points Risk Reduction Delta** (**52.21% relative risk reduction**).
   - **Online Backup** (+8.31 pts) and **Device Protection** (+6.58 pts) also deliver strong defensive value.

   - Subscribing to `Streaming Movies` (-4.38 pts) or `Streaming TV` (-5.60 pts) is associated with higher churn rates (~29–30%).

PS6: Financial Risk & Revenue Loss Quantification

### Business Question
What is the exact financial magnitude of customer churn across Monthly Recurring Revenue (MRR), annualized run-rate exposure, ARPU discrepancies, and realized Customer Lifetime Value (CLTV)?

### Why This Matters
For executive leadership and the CFO, customer headcount churn matters primarily through its impact on cash flow. Understanding whether departing accounts are low-margin bargain hunters or high-value premium subscribers dictates the commercial urgency and payback economics of retention investments.

### Analytical Method
1. Calculate portfolio-wide and segment revenue totals:
   - **Total Monthly Recurring Revenue (MRR)**: $\text{MRR}_{total} = \sum_{i=1}^N \text{Monthly Charge}_i$
   - **Monthly Revenue Retained**: $\text{MRR}_{retained} = \sum_{i \in \text{Retained}} \text{Monthly Charge}_i$
   - **Monthly Revenue Lost to Churn**: $\text{MRR}_{lost} = \sum_{i \in \text{Churned}} \text{Monthly Charge}_i$
   - **Revenue Exposure Share (%)**: $\frac{\text{MRR}_{lost}}{\text{MRR}_{total}} \times 100$
   - **Illustrative Annualized Run-Rate Exposure**: $\text{ARR Exposure} = \text{MRR}_{lost} \times 12$
2. Evaluate per-account realization metrics:
   - Average Monthly Charge (ARPU) for Churned vs. Retained accounts
   - Realized Cumulative Charges ($\mu(\text{Total Charges})$)
   - Unrealized Lifetime Value Deficit
3. Construct dual financial visualizations:
   - **Plot 12**: Monthly Recurring Revenue (MRR) Bridge / Waterfall Chart detailing portfolio retention versus active loss.
   - **Plot 13**: Comparative Financial Realization Gap Bar Chart contrasting ARPU, Cumulative Charges, and CLTV.

````python
# PS6: Financial Risk & Revenue Loss Quantification Analysis
total_monthly_rev = df_clean['Monthly Charge'].sum()
churned_monthly_rev = df_clean[df_clean['Churn'] == 1]['Monthly Charge'].sum()
retained_monthly_rev = df_clean[df_clean['Churn'] == 0]['Monthly Charge'].sum()
pct_monthly_rev_lost = (churned_monthly_rev / total_monthly_rev) * 100

avg_monthly_churned = df_clean[df_clean['Churn'] == 1]['Monthly Charge'].mean()
avg_monthly_retained = df_clean[df_clean['Churn'] == 0]['Monthly Charge'].mean()
arpu_gap = avg_monthly_churned - avg_monthly_retained

avg_total_charges_churned = df_clean[df_clean['Churn'] == 1]['Total Charges'].mean()
avg_total_charges_retained = df_clean[df_clean['Churn'] == 0]['Total Charges'].mean()

avg_cltv_churned = df_clean[df_clean['Churn'] == 1]['CLTV'].mean()
avg_cltv_retained = df_clean[df_clean['Churn'] == 0]['CLTV'].mean()

annualized_loss_exposure = churned_monthly_rev * 12

print("=" * 76)
print("PS6: FINANCIAL RISK & REVENUE LOSS QUANTIFICATION SUMMARY")
print("=" * 76)
print(f"Total Monthly Recurring Revenue (MRR)   : ${total_monthly_rev:,.2f}")
print(f"Retained Monthly Recurring Revenue      : ${retained_monthly_rev:,.2f} ({100 - pct_monthly_rev_lost:.2f}%)")
print(f"Churned Monthly Revenue Lost            : ${churned_monthly_rev:,.2f} ({pct_monthly_rev_lost:.2f}%)")
print(f"Illustrative Annualized Run-Rate Exposure: ${annualized_loss_exposure:,.2f}")
print("-" * 76)
print(f"Average Monthly Charge — Churned Customers : ${avg_monthly_churned:.2f}")
print(f"Average Monthly Charge — Retained Customers: ${avg_monthly_retained:.2f}")
print(f"ARPU Gap (Premium Churn Premium)           : +${arpu_gap:.2f} (+{((avg_monthly_churned / avg_monthly_retained) - 1)*100:.1f}%)")
print("-" * 76)
print(f"Average Cumulative Charges Realized (Churned) : ${avg_total_charges_churned:,.2f}")
print(f"Average Cumulative Charges Realized (Retained): ${avg_total_charges_retained:,.2f}")
print(f"Unrealized Lifetime Charge Deficit per Churn  : -${avg_total_charges_retained - avg_total_charges_churned:,.2f}")
print(f"Average Predicted CLTV (Churned vs. Retained) : ${avg_cltv_churned:,.2f} vs. ${avg_cltv_retained:,.2f}")
print("=" * 76)

# Figure 1: Plot 12 — Monthly Revenue Bridge / Waterfall Chart
fig_wf, ax1 = plt.subplots(figsize=(8, 5.2))
wf_labels = ['Total Portfolio MRR', 'Retained Active MRR', 'Lost Churned MRR']
wf_values = [total_monthly_rev, retained_monthly_rev, churned_monthly_rev]
wf_colors = [BI_PALETTE['primary'], BI_PALETTE['safe'], BI_PALETTE['churn']]

bars1 = ax1.bar(wf_labels, wf_values, color=wf_colors, width=0.45, edgecolor='black', linewidth=0.8)
for i, bar in enumerate(bars1):
    h = bar.get_height()
    pct = (h / total_monthly_rev) * 100
    ax1.annotate(f'${h:,.0f}\n({pct:.1f}%)', xy=(bar.get_x() + bar.get_width() / 2, h),
                 xytext=(0, 6), textcoords='offset points', ha='center', va='bottom',
                 fontsize=11, fontweight='bold', color='#2c3e50')

ax1.set_title('Plot 12: Monthly Recurring Revenue (MRR) Distribution: Retained vs. Lost', fontsize=12, fontweight='bold', pad=12, color=BI_PALETTE['primary'])
ax1.set_ylabel('Monthly Recurring Revenue ($)', fontsize=10, fontweight='bold')
ax1.set_ylim(0, total_monthly_rev * 1.20)
ax1.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

# Figure 2: Plot 13 — Financial Realization Gap Bar Chart
fig_gap, ax2 = plt.subplots(figsize=(8.5, 5.2))
arpu_metrics = ['Monthly Charge ($)', 'Total Charges ($ / 10)', 'CLTV ($ / 10)']
churned_vals = [avg_monthly_churned, avg_total_charges_churned / 10, avg_cltv_churned / 10]
retained_vals = [avg_monthly_retained, avg_total_charges_retained / 10, avg_cltv_retained / 10]

x = np.arange(len(arpu_metrics))
w = 0.35
b1 = ax2.bar(x - w/2, retained_vals, w, label='Retained Customers', color=BI_PALETTE['retained'], edgecolor='black', linewidth=0.8)
b2 = ax2.bar(x + w/2, churned_vals, w, label='Churned Customers', color=BI_PALETTE['churn'], edgecolor='black', linewidth=0.8)

for i in range(len(arpu_metrics)):
    orig_r = [avg_monthly_retained, avg_total_charges_retained, avg_cltv_retained][i]
    orig_c = [avg_monthly_churned, avg_total_charges_churned, avg_cltv_churned][i]
    ax2.annotate(f'${orig_r:,.1f}', xy=(x[i] - w/2, retained_vals[i]), xytext=(0, 4), textcoords='offset points', ha='center', fontsize=10, fontweight='bold')
    ax2.annotate(f'${orig_c:,.1f}', xy=(x[i] + w/2, churned_vals[i]), xytext=(0, 4), textcoords='offset points', ha='center', fontsize=10, fontweight='bold')

ax2.set_xticks(x)
ax2.set_xticklabels(arpu_metrics, fontsize=10, fontweight='bold')
ax2.set_title('Plot 13: Financial Realization Gap: Retained vs. Churned Accounts', fontsize=12, fontweight='bold', pad=12, color=BI_PALETTE['primary'])
ax2.set_ylabel('Normalized Metric Value ($)', fontsize=10, fontweight='bold')
ax2.set_ylim(0, max(retained_vals + churned_vals) * 1.24)
ax2.legend(loc='upper right')
ax2.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
````

## PS7: Demographic Profiling & Correlation Matrix

### Business Question
How does customer attrition vary across demographic subgroups, and what are the primary positive and negative independent statistical drivers of customer churn?

### Why This Matters
Demographic segmentation enables customer care and marketing teams to tailor messaging and support channels. Furthermore, correlation analysis isolates pre-churn operational drivers from post-churn diagnostic variables, providing a clean foundation for executive decision-making.

### Analytical Method
1. Evaluate bivariate churn across key demographic attributes:
   - `Senior Citizen` (0 vs. 1)
   - `Gender` (Female vs. Male)
   - `Partner` (0 vs. 1)
   - `Dependents` (0 vs. 1)
2. One-hot encode multi-class categorical features (`Contract`, `Payment Method`, `Internet Type`, `Offer`, `Gender`).
3. Compute Pearson correlation coefficients ($r$) against target variable `Churn`.
4. **Isolate Post-Event Feature Leakage**: Review and exclude post-churn outcomes (`Satisfaction Score`, `Churn Score`, `Churn Reason`, `Churn Category`, `Customer Status`) from independent driver rankings.
5. Render three visual perspectives:
   - **Plot 14**: Demographic Churn Rate Comparison Bar Chart.
   - **Plot 15**: Annotated Pearson Correlation Heatmap of top features with Churn.
   - **Plot 16**: Horizontal Bar Chart of top independent positive and negative churn drivers.

````python
# PS7: Demographic Profiling & Correlation Matrix Analysis
demo_summary = []
for col, name in [('Senior Citizen', 'Senior Citizen Status'), ('Gender', 'Gender'), ('Partner', 'Partner Status'), ('Dependents', 'Dependents Status')]:
    for val in sorted(df_clean[col].unique()):
        sub = df_clean[df_clean[col] == val]
        cnt = len(sub)
        churn_cnt = int(sub['Churn'].sum())
        rt = (churn_cnt / cnt) * 100
        val_label = 'Yes' if val == 1 else ('No' if val == 0 else str(val))
        demo_summary.append({
            'Demographic Attribute': name,
            'Subgroup': val_label,
            'Customer Count': cnt,
            'Churned Count': churn_cnt,
            'Churn Rate (%)': rt
        })
demo_df = pd.DataFrame(demo_summary)

print("=" * 80)
print("PS7: DEMOGRAPHIC PROFILING & CHURN RATE BREAKDOWN")
print("=" * 80)
print(demo_df.round(2).to_string(index=False))
print("=" * 80)

# Feature Encoding for Correlation Matrix
cat_features = ['Gender', 'Contract', 'Payment Method', 'Internet Type', 'Offer']
df_encoded = pd.get_dummies(df_clean, columns=cat_features, drop_first=False)

# Explicitly isolate post-event diagnostic / leakage columns
post_event_leakage_cols = ['Satisfaction Score', 'Churn Score', 'Churn Category', 'Churn Reason', 'Customer Status']
admin_cols = ['Customer ID', 'City', 'Country', 'State', 'Quarter', 'Lat Long', 'Tenure_Cohort']

# Clean numeric matrix for correlation
df_corr_all = df_encoded.drop(columns=[c for c in admin_cols if c in df_encoded.columns])
df_corr_all = df_corr_all.select_dtypes(include=[np.number, bool])
for c in df_corr_all.columns:
    if df_corr_all[c].dtype == bool:
        df_corr_all[c] = df_corr_all[c].astype(int)

# Pearson correlation across all features
all_corr = df_corr_all.corr()['Churn'].drop('Churn').sort_values(ascending=False)

# Filter for INDEPENDENT pre-churn drivers (excluding leakage variables)
independent_corr = df_corr_all.drop(columns=[c for c in post_event_leakage_cols if c in df_corr_all.columns]).corr()['Churn'].drop('Churn').sort_values(ascending=False)
top_pos_independent = independent_corr.head(8)
top_neg_independent = independent_corr.tail(8).sort_values(ascending=True)

print("\nTop Pre-Event Independent Drivers Increasing Churn (Accelerators):")
for feat, val in top_pos_independent.items():
    print(f"  + {feat:<32} : r = {val:+.4f}")

print("\nTop Pre-Event Independent Drivers Dampening Churn (Protectors):")
for feat, val in top_neg_independent.items():
    print(f"  - {feat:<32} : r = {val:+.4f}")
print("=" * 80)

# Multi-panel Visual Architecture (Plots 14, 15, 16)
# Figure 1: Demographic Comparison (Plot 14)
fig_demo, ax_d = plt.subplots(figsize=(11, 4.8))
demo_labels = [f"{r['Demographic Attribute'][:12]}: {r['Subgroup']}" for _, r in demo_df.iterrows()]
demo_rates = demo_df['Churn Rate (%)'].values
demo_counts = demo_df['Customer Count'].values
demo_colors = [BI_PALETTE['churn'] if r > churn_rate_pct else BI_PALETTE['safe'] for r in demo_rates]

bars_d = ax_d.bar(demo_labels, demo_rates, color=demo_colors, width=0.52, edgecolor='black', linewidth=0.8)
ax_d.axhline(churn_rate_pct, color='#2c3e50', linestyle='--', linewidth=1.5, label=f'Baseline Churn ({churn_rate_pct:.1f}%)')
for i, bar in enumerate(bars_d):
    h = bar.get_height()
    n = demo_counts[i]
    ax_d.annotate(f'{h:.1f}%\n(N={n:,})', xy=(bar.get_x() + bar.get_width() / 2, h),
                  xytext=(0, 4), textcoords='offset points', ha='center', va='bottom',
                  fontsize=9, fontweight='bold', color='#2c3e50')

ax_d.set_title('Plot 14: Customer Churn Rate by Demographic Subgroup vs. Portfolio Baseline', fontsize=12, fontweight='bold', pad=12, color=BI_PALETTE['primary'])
ax_d.set_ylabel('Cohort Churn Rate (%)', fontsize=10, fontweight='bold')
ax_d.set_ylim(0, 48)
ax_d.set_xticklabels(demo_labels, rotation=25, ha='right', fontsize=9, fontweight='bold')
ax_d.legend(loc='upper right')
ax_d.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()

# Figure 2: Correlation Heatmap (Plot 15) & Independent Drivers (Plot 16)
fig_corr, (ax_h, ax_b) = plt.subplots(1, 2, figsize=(17, 7.5))
plt.subplots_adjust(left=0.08, right=0.95, top=0.90, bottom=0.12, wspace=0.32)

# Plot 15: Heatmap of Top Features
top_heatmap_feats = pd.concat([top_pos_independent.head(6), top_neg_independent.head(6)]).index.tolist()
corr_matrix_sub = df_corr_all[top_heatmap_feats + ['Churn']].corr()

sns.heatmap(corr_matrix_sub[['Churn']].sort_values('Churn', ascending=False),
            annot=True, fmt='.3f', cmap='coolwarm', cbar_kws={'label': 'Pearson Correlation (r)'},
            ax=ax_h, vmin=-0.8, vmax=0.8, linewidths=0.6, annot_kws={'size': 10, 'weight': 'bold'})
ax_h.set_title('Plot 15: Pearson Correlation of Pre-Event Features with Churn', fontsize=12, fontweight='bold', pad=12, color=BI_PALETTE['primary'])

# Plot 16: Horizontal Bar Chart of Drivers
driver_names = top_pos_independent.index.tolist()[::-1] + top_neg_independent.index.tolist()
driver_vals = top_pos_independent.values.tolist()[::-1] + top_neg_independent.values.tolist()
driver_colors = [BI_PALETTE['churn'] if v > 0 else BI_PALETTE['safe'] for v in driver_vals]

bars_drv = ax_b.barh(driver_names, driver_vals, color=driver_colors, edgecolor='black', linewidth=0.6, height=0.6)
ax_b.axvline(0, color='black', linewidth=1)
for bar in bars_drv:
    w = bar.get_width()
    xpos = w + (0.015 if w >= 0 else -0.015)
    ha = 'left' if w >= 0 else 'right'
    sign = '+' if w > 0 else ''
    ax_b.annotate(f'{sign}{w:.3f}', xy=(xpos, bar.get_y() + bar.get_height() / 2),
                  va='center', ha=ha, fontsize=9, fontweight='bold', color='#2c3e50')

ax_b.set_title('Plot 16: Top Independent Drivers of Churn (Accelerators vs. Protectors)', fontsize=12, fontweight='bold', pad=12, color=BI_PALETTE['primary'])
ax_b.set_xlabel('Pearson Correlation Coefficient (r)', fontsize=10, fontweight='bold')
ax_b.set_xlim(-0.45, 0.55)
ax_b.grid(axis='x', alpha=0.3)
plt.show()
````
